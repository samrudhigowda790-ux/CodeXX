```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

```python
np.random.seed(42)
```

```python
n, p = 20, 0.35
binom = stats.binom (n,p)
x = np.arange (0, n+1)
```

```python
print (f"Mean = {binom.mean():.4f} (Theory: np={n*p})")
```

```python
binom.var()
```

```python
binom.moment(2)
```

```python
binom.stats("s")
```

```python
binom.cdf(10)
```

```python
binom.pmf(7)
```
